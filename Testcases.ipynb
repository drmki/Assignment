```python

import unittest

# ============================================================
# CUSTOM TEST RESULT
# Displays test title, user inputs and PASS/FAIL result.
# ============================================================

class CustomTestResult(unittest.TextTestResult):

    def startTest(self, test):
        super().startTest(test)

        print("\n" + "=" * 65)
        print("STARTING:", test.shortDescription())
        print("=" * 65)

    def addSuccess(self, test):
        super().addSuccess(test)

        print("\nRESULT: PASS")
        print("MESSAGE: Test case successfully passed.")

    def addFailure(self, test, err):
        super().addFailure(test, err)

        print("\nRESULT: FAIL")
        print("MESSAGE: Test case failed. Input validation or expected result did not match.")

    def addError(self, test, err):
        super().addError(test, err)

        print("\nRESULT: ERROR")
        print("MESSAGE: An unexpected error occurred.")

    def addSkip(self, test, reason):
        super().addSkip(test, reason)

        print("\nRESULT: SKIPPED")
        print("MESSAGE:", reason)


class CustomTestRunner(unittest.TextTestRunner):
    resultclass = CustomTestResult


# ============================================================
# USER INPUT HELPER FUNCTIONS
# ============================================================

def get_user_input(message):
    return input(message).strip()


def get_integer_input(message):
    while True:
        try:
            return int(input(message))
        except ValueError:
            print("Invalid input! Please enter a valid integer.")


def get_float_input(message):
    while True:
        try:
            return float(input(message))
        except ValueError:
            print("Invalid input! Please enter a valid number.")


# ============================================================
# EXECUTE ALL TEST CASES
# ============================================================

if __name__ == "__main__":

    print("\n")
    print("=" * 65)
    print("       ROYAL CARE CLINIC - APPLICATION TESTING")
    print("=" * 65)

    print("\nPlease enter the required information for each test case.")

    suite = unittest.defaultTestLoader.loadTestsFromTestCase(
        TestRoyalCareClinic
    )

    runner = CustomTestRunner(
        verbosity=0,
        resultclass=CustomTestResult
    )

    result = runner.run(suite)

    # ========================================================
    # FINAL TEST REPORT
    # ========================================================

    total = result.testsRun
    failed = len(result.failures)
    errors = len(result.errors)
    skipped = len(result.skipped)
    passed = total - failed - errors - skipped

    print("\n")
    print("=" * 65)
    print("       ROYAL CARE CLINIC - FINAL TEST REPORT")
    print("=" * 65)

    print("Total Test Cases        :", total)
    print("Successfully Passed     :", passed)
    print("Failed Test Cases       :", failed)
    print("Errors                  :", errors)
    print("Skipped                 :", skipped)

    print("-" * 65)

    if failed == 0 and errors == 0:
        print("FINAL STATUS: ALL TEST CASES SUCCESSFULLY PASSED!")
    else:
        print("FINAL STATUS: SOME TEST CASES FAILED.")

    print("=" * 65)
```
